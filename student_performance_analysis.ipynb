# Student Performance Analysis (UCI Dataset)
Original analytics project using real data from the UCI Machine Learning Repository.


```python
import pandas as pd
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00320/student-mat.csv'
data = pd.read_csv(url, sep=';', engine='python')
data.head()
```

```python
# Basic dataset information
print(data.shape)
print(data.describe())
```

```python
import matplotlib.pyplot as plt
plt.figure()
plt.scatter(data['studytime'], data['G3'])
plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade')
plt.show()
```
