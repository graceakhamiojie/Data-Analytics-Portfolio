# Bank Marketing Campaign Analysis (UCI Dataset)
Original analytics project using real data from the UCI Machine Learning Repository.


```python
import pandas as pd
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00222/bank.csv'
data = pd.read_csv(url, sep=';', engine='python')
data.head()
```

```python
# Basic dataset information
print(data.shape)
print(data.describe())
```

```python
import matplotlib.pyplot as plt
plt.figure()
data['y'].value_counts().plot(kind='bar')
plt.title('Marketing Campaign Outcomes')
plt.xlabel('Outcome')
plt.ylabel('Number of Customers')
plt.show()
```
